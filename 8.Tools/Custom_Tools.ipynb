```python
from langchain_core.tools import tool
```

## **USING @BOOL DECORATOR**

```python
# step 1 : build you logic
def multiply(a , b):
  return a * b
```

```python
# step2 : add type hints
def multiply(a :int , b :int) -> int :
  return a * b
```

```python
# step3 : add tool decorator
@tool
def multiply(a :int , b :int) -> int :
  """Multiplying two number"""
  return a * b
```

```python
result = multiply.invoke({"a" : 10 , "b" : 20})
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

### **USING STRUCTURED TOOL**

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel , Field
```

```python
class Add_input(BaseModel):
  a : int = Field(required = True , description= "The first number to add")
  b : int = Field(required = True , description= "The second number to add")
```

```python
def add(a :int , b :int) -> int :
  """Adding two number"""
  return a + b
```

```python
add_tool = StructuredTool.from_function(
    func = add,
    name = "addition",
    description = "addition of two number",
    args_schema = Add_input
)
```

```python
result = add_tool.invoke({"a" : 10 , "b" : 15})
print(result)
```

```python
print(add_tool.name)
print(add_tool.description)
print(add_tool.args)
```

### **USING BASE TOOL**

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
class Subtraction_input(BaseModel):
  a : int = Field(required = True , description= "The first number to sub")
  b : int = Field(required = True , description= "The second number to sub")
```

```python
class Subtraction(BaseTool):
  name : str = "substraction",
  description : str = "It substract two numbers",
  args_schema  : Type[BaseModel] = Subtraction_input

  def _run(self , a : int , b : int) -> int:
    return a - b
```

```python
sub_tool = Subtraction()
```

```python
result = sub_tool.invoke({"a" : 30 , "b" : 20})
print(result)
```

```python
print(sub_tool.name)
print(sub_tool.description)
```
