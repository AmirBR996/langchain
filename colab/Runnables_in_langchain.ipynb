```python
from abc import ABC, abstractmethod
```

```python
class Runnable(ABC):
  @abstractmethod
  def invoke(input_data):
    pass
```

```python
import random
class demollm(Runnable):
  def __init__(self):
    print("LLM created")
  def invoke(self , prompt):
    response_list =[
        'kathmandu is the capital of nepal',
        'npl is a cricket leauge',
        'dl stands for deep learning'
    ]
    return {'response' : random.choice(response_list)}
  def predict(self , prompt):
    response_list =[
        'kathmandu is the capital of nepal',
        'npl is a cricket leauge',
        'dl stands for deep learning'
    ]
    return {'response' : random.choice(response_list)}
```

```python
class demoPrompttemplate(Runnable):
  def __init__(self , template , input_variables):
    self.template = template
    self.input_varaibles = input_variables
  def invoke(self,input_dict):
    return self.template.format(**input_dict)
  def format(self,input_dict):
    return self.template.format(**input_dict)
```

```python
class runnable_connector(Runnable):
  def __init__(self , runnable_list):
    self.runnable_list = runnable_list
  def invoke(self,input_data):
    for runnable in self.runnable_list:
      input_data = runnable.invoke(input_data)
    return input_data
```

```python
llm = demollm()
```

```python
template = demoPrompttemplate(
    template = 'Write a {length} note about {topic}',
    input_variables=['length' , 'topic']
)
```

```python
chain = runnable_connector([template , llm])
```

```python
chain.invoke({'length' : 'short' , 'topic' : 'quantam computing'})
```
