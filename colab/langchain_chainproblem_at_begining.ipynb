```python
import random
```

```python
class demollm:
  def __init__(self):
    print("LLM created")
  def predict(self , prompt):
    response_list =[
        'kathmandu is the capital of nepal',
        'npl is a cricket leauge',
        'dl stands for deep learning'
    ]
    return {'response' : random.choice(response_list)}
```

```python
class demoPrompttemplate:
  def __init__(self , template , input_variables):
    self.template = template
    self.input_varaibles = input_variables
  def format(self,input_dict):
    return self.template.format(**input_dict)
```

```python
template = demoPrompttemplate(
    template = 'Write a {length} note about {topic}',
    input_variables=['length' , 'topic']
)
```

```python
prompt = template.format({'length' : 'short' , 'topic' : 'quantam computing'})
```

```python
llm = demollm()
```

```python
llm.predict(prompt)
```

```python
class demochain():
  def __init__(self , llm , prompt):
    self.llm = llm
    self.prompt = prompt
  def run(self,input_dict):
    final_prompt = self.prompt.format(input_dict)
    result = self.llm.predict(final_prompt)
    return result['response']
```

```python
template = demoPrompttemplate(
    template = 'Write a {length} note about {topic}',
    input_variables=['length' , 'topic']
)
```

```python
chain = demochain(llm,template)
```

```python
chain.run({'length' : 'long' , 'topic' : 'india'})
# it is not that flexible to create any type of task
```
